```python
from dataclasses import dataclass

import numpy as np
import pandas as pd
import polars as pl


@dataclass
class XYZ:
    x: float
    y: float
    z: float


data = {
    "x": [7.7, 4.3],
    "y": [5.2, 6.5],
    "z": [9.7, 0.7]
}

scalar = {k: v[0] for k, v in data.items()}

constructors = {
    "np_array": np.array,
    "pd_series": pd.Series,
    "pl_series": pl.Series,
}

inputs = {
    **{
        name: {k: constructor(v) for k, v in data.items()}
        for name, constructor in constructors.items()
    },
    "number": scalar,
    "pd_df": pd.DataFrame(data),
    "pl_df": pl.DataFrame(data),
    "pd_series_dict": pd.Series(scalar),
    "dataclass": XYZ(**scalar),
}
```

```python
from mi77_utils.computations import Calculator

from math import sqrt


def transform(normalized):
    return normalized * 100


def validate(processed):
    return processed <= 100


calc = Calculator(
    squared_norm="x**2 + y**2",
    norm=lambda squared_norm: sqrt(squared_norm),
    normalized=lambda norm, z: norm / z,
    processed=transform,
    valid=validate,
)

for name, values in inputs.items():
    display(name, calc.calculate(values))
```
